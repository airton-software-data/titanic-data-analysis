**Importando as bibliotecas**

```python
import pandas as pd # manipulação de dados em tabelas
import matplotlib.pyplot as plt # visualização de gráficos
import seaborn as sns # visualização de gráficos
```

```python
# Carregar o dataset Titanic (disponível no Kaggle ou seaborn)
df = sns.load_dataset("titanic")

# Visualizar as primeiras linhas
print(df.head())
```

```python
# Estrutura do dataset mostra colunas, tipos de dados e valores nulos.
print(df.info())
```

```python
# Estatísticas descritivas estatísticas básicas (média, desvio padrão, etc.).
print(df.describe())
```

**Análise exploratória**

```python
# Distribução de sobreviventes
sns.countplot(x="survived", data=df)
plt.title("Distribuição de Sobreviventes")
plt.show()

# Relação entre sexo e sobrevivência
sns.countplot(x="sex", hue="survived", data=df)
plt.title("Sobrevivência por sexo")
plt.show()

# Idade vs sobrevivência
sns.histplot(df["age"], bins=30, kde=True)
plt.title("Distribuição de idades")
plt.show()
```
